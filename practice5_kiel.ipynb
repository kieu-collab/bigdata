```python
# ================================================================
# BIG DATA PERFORMANCE BENCHMARK ON REAL PARQUET FILE
#
# Compare:
#   1. Pandas
#   2. Polars
#   3. DuckDB
#   4. Vaex
#   5. Dask
#   6. Ray Data
#   7. PySpark
#
# Workload:
#   Parquet
#      ↓
#   Select columns
#      ↓
#   Filter
#      ↓
#   GroupBy
#      ↓
#   COUNT + SUM + MEAN
#      ↓
#   Materialize result
#
# Output:
#   - First 10 rows
#   - Parquet metadata
#   - Auto-selected analysis columns
#   - Execution times
#   - Ranking
#   - Speedup vs Pandas
#   - Validation
#   - Chart
#   - CSV results
# ================================================================


# ================================================================
# 0. INSTALL PACKAGES
# ================================================================

import os
import sys
import subprocess
import warnings

warnings.filterwarnings("ignore")

print("=" * 80)
print("INSTALLING BIG-DATA LIBRARIES")
print("=" * 80)

MAIN_PACKAGES = [
    "pandas",
    "numpy",
    "pyarrow",
    "polars",
    "duckdb",
    "dask[dataframe]",
    "ray[data]",
    "pyspark",
    "psutil",
    "matplotlib",
]

subprocess.check_call(
    [sys.executable, "-m", "pip", "install", "-q"] + MAIN_PACKAGES
)

# Vaex is installed separately.
# If Vaex cannot be installed in the current Colab runtime,
# the rest of the benchmark will still run.
print("\nInstalling Vaex...")

vaex_install = subprocess.run(
    [
        sys.executable,
        "-m",
        "pip",
        "install",
        "-q",
        "vaex-core"
    ],
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE,
    text=True
)

if vaex_install.returncode == 0:
    print("Vaex installed.")
else:
    print("Vaex installation was not successful.")
    print("Benchmark will continue without Vaex.")


# ================================================================
# 1. IMPORT LIBRARIES
# ================================================================

import gc
import glob
import time
import random
import platform

import numpy as np
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq

import polars as pl
import duckdb
import dask.dataframe as dd

import matplotlib.pyplot as plt

from IPython.display import display


# ================================================================
# 2. CONFIGURATION
# ================================================================

# ------------------------------------------------
# OPTION A:
# Put a .parquet file directly in /content
# and leave PARQUET_PATH = None
# ------------------------------------------------

PARQUET_PATH = None


# ------------------------------------------------
# OPTION B:
# If using Google Drive, uncomment:
#
# from google.colab import drive
# drive.mount("/content/drive")
#
# PARQUET_PATH = (
#     "/content/drive/MyDrive/BigData/data.parquet"
# )
# ------------------------------------------------


# ------------------------------------------------
# Manual columns - OPTIONAL
#
# Leave None for automatic detection.
#
# Example:
# GROUP_COL  = "Country"
# VALUE_COL  = "Revenue"
# FILTER_COL = "Quantity"
# ------------------------------------------------

GROUP_COL = None
VALUE_COL = None
FILTER_COL = None


# Number of repeated benchmark runs.
N_RUNS = 3

RANDOM_SEED = 2026

CPU_COUNT = os.cpu_count() or 2


print("\n" + "=" * 80)
print("SYSTEM")
print("=" * 80)

print("Python        :", sys.version.split()[0])
print("CPU cores     :", CPU_COUNT)
print("Platform      :", platform.platform())
print("Benchmark runs:", N_RUNS)


# ================================================================
# 3. FIND / UPLOAD PARQUET FILE
# ================================================================

if PARQUET_PATH is None:

    candidates = sorted(
        glob.glob("/content/*.parquet")
    )

    if candidates:
        PARQUET_PATH = candidates[0]

        print("\nAutomatically detected:")
        print(PARQUET_PATH)

    else:
        print("\nNo Parquet file found in /content.")
        print("Please select a .parquet file.")

        from google.colab import files

        uploaded = files.upload()

        parquet_names = [
            name
            for name in uploaded.keys()
            if name.lower().endswith(".parquet")
        ]

        if not parquet_names:
            raise RuntimeError(
                "No .parquet file was uploaded."
            )

        PARQUET_PATH = os.path.abspath(
            parquet_names[0]
        )


if not os.path.exists(PARQUET_PATH):
    raise FileNotFoundError(
        f"File not found: {PARQUET_PATH}"
    )


print("\n" + "=" * 80)
print("PARQUET FILE")
print("=" * 80)

print("Path:",
      PARQUET_PATH)

FILE_SIZE_MB = (
    os.path.getsize(PARQUET_PATH)
    / 1024
    / 1024
)

print(
    f"File size: {FILE_SIZE_MB:,.2f} MB"
)


# ================================================================
# 4. PARQUET METADATA
# ================================================================

parquet_file = pq.ParquetFile(
    PARQUET_PATH
)

metadata = parquet_file.metadata
schema = parquet_file.schema_arrow

NUM_ROWS = metadata.num_rows
NUM_COLS = metadata.num_columns
NUM_ROW_GROUPS = metadata.num_row_groups


print("\n" + "=" * 80)
print("PARQUET METADATA")
print("=" * 80)

print(
    f"Rows       : {NUM_ROWS:,}"
)

print(
    f"Columns    : {NUM_COLS:,}"
)

print(
    f"Row groups : {NUM_ROW_GROUPS:,}"
)

print(
    f"File size  : {FILE_SIZE_MB:,.2f} MB"
)


# ================================================================
# 5. DISPLAY SCHEMA
# ================================================================

schema_table = pd.DataFrame(
    {
        "Column": [
            field.name
            for field in schema
        ],

        "Data type": [
            str(field.type)
            for field in schema
        ]
    }
)

print("\n" + "=" * 80)
print("DATA SCHEMA")
print("=" * 80)

display(schema_table)


# ================================================================
# 6. LOAD ONLY A SMALL SAMPLE
#    IMPORTANT:
#    We do NOT load the entire dataset just to preview it.
# ================================================================

if NUM_ROWS == 0:
    raise RuntimeError(
        "Parquet file contains zero rows."
    )


sample_size = min(
    100_000,
    NUM_ROWS
)

sample_batch = next(
    parquet_file.iter_batches(
        batch_size=sample_size
    )
)

sample_df = (
    sample_batch
    .to_pandas()
)


# ================================================================
# 7. SHOW FIRST 10 ROWS
# ================================================================

print("\n" + "=" * 80)
print("FIRST 10 ROWS")
print("=" * 80)

display(
    sample_df.head(10)
)


# ================================================================
# 8. IDENTIFY NUMERIC / CATEGORICAL COLUMNS
# ================================================================

numeric_cols = []

categorical_cols = []

for field in schema:

    dtype = field.type

    if (
        pa.types.is_integer(dtype)
        or pa.types.is_floating(dtype)
        or pa.types.is_decimal(dtype)
    ):
        numeric_cols.append(
            field.name
        )

    elif (
        pa.types.is_string(dtype)
        or pa.types.is_large_string(dtype)
        or pa.types.is_dictionary(dtype)
        or pa.types.is_boolean(dtype)
    ):
        categorical_cols.append(
            field.name
        )


print("\nNumeric columns:")
print(numeric_cols)

print("\nCategorical columns:")
print(categorical_cols)


if len(numeric_cols) == 0:
    raise RuntimeError(
        """
No numeric column was detected.

This benchmark needs at least one numeric
column for SUM / MEAN / filtering.
"""
    )


# ================================================================
# 9. HELPER FUNCTIONS FOR AUTOMATIC COLUMN SELECTION
# ================================================================

def sample_nunique(column):

    try:
        return int(
            sample_df[column]
            .nunique(dropna=True)
        )

    except Exception:
        return 10**12


def choose_by_keyword(
    candidates,
    keywords
):

    if not candidates:
        return None

    for keyword in keywords:

        for column in candidates:

            if keyword in column.lower():
                return column

    return None


# Common business-data keywords.

GROUP_HINTS = [
    "country",
    "category",
    "segment",
    "region",
    "state",
    "city",
    "department",
    "product",
    "brand",
    "channel",
    "status",
    "type",
    "class",
    "group"
]

VALUE_HINTS = [
    "revenue",
    "sales",
    "amount",
    "total",
    "profit",
    "income",
    "price",
    "value",
    "cost",
    "quantity",
    "qty"
]

FILTER_HINTS = [
    "quantity",
    "qty",
    "revenue",
    "sales",
    "amount",
    "price",
    "value",
    "profit",
    "score",
    "rating",
    "age"
]


# ================================================================
# 10. AUTO-DETECT GROUP COLUMN
# ================================================================

if GROUP_COL is None:

    # Prefer categorical fields with reasonable
    # cardinality.

    reasonable_cat = [

        c for c in categorical_cols

        if (
            2
            <= sample_nunique(c)
            <= 5_000
        )
    ]

    GROUP_COL = choose_by_keyword(
        reasonable_cat,
        GROUP_HINTS
    )

    if GROUP_COL is None and reasonable_cat:

        GROUP_COL = min(
            reasonable_cat,
            key=sample_nunique
        )


    # If no suitable text/category column,
    # try a low-cardinality numeric column.

    if GROUP_COL is None:

        numeric_group_candidates = [

            c for c in numeric_cols

            if (
                2
                <= sample_nunique(c)
                <= 1_000
            )
        ]

        GROUP_COL = choose_by_keyword(
            numeric_group_candidates,
            GROUP_HINTS
        )

        if (
            GROUP_COL is None
            and numeric_group_candidates
        ):

            GROUP_COL = min(
                numeric_group_candidates,
                key=sample_nunique
            )


    # Last fallback.

    if GROUP_COL is None:

        if categorical_cols:

            GROUP_COL = min(
                categorical_cols,
                key=sample_nunique
            )

        else:

            GROUP_COL = min(
                numeric_cols,
                key=sample_nunique
            )


# ================================================================
# 11. AUTO-DETECT VALUE COLUMN
# ================================================================

if VALUE_COL is None:

    VALUE_COL = choose_by_keyword(
        numeric_cols,
        VALUE_HINTS
    )

    if VALUE_COL is None:

        VALUE_COL = numeric_cols[0]


# ================================================================
# 12. AUTO-DETECT FILTER COLUMN
# ================================================================

if FILTER_COL is None:

    filter_candidates = [

        c for c in numeric_cols

        if c != VALUE_COL
    ]

    FILTER_COL = choose_by_keyword(
        filter_candidates,
        FILTER_HINTS
    )

    if FILTER_COL is None:

        if filter_candidates:
            FILTER_COL = filter_candidates[0]

        else:
            FILTER_COL = VALUE_COL


# ================================================================
# 13. CHECK USER / AUTO COLUMNS
# ================================================================

all_columns = [
    field.name
    for field in schema
]

for c in [
    GROUP_COL,
    VALUE_COL,
    FILTER_COL
]:

    if c not in all_columns:

        raise ValueError(
            f"Column not found: {c}"
        )


print("\n" + "=" * 80)
print("BENCHMARK COLUMNS")
print("=" * 80)

print(
    "GROUP BY column :",
    GROUP_COL
)

print(
    "VALUE column    :",
    VALUE_COL
)

print(
    "FILTER column   :",
    FILTER_COL
)

print(
    "Sample groups   :",
    f"{sample_nunique(GROUP_COL):,}"
)


# ================================================================
# 14. DETERMINE FILTER THRESHOLD
#
# Example:
#
# FILTER_COL > median
# ================================================================

filter_sample = pd.to_numeric(
    sample_df[FILTER_COL],
    errors="coerce"
)

FILTER_THRESHOLD = float(
    filter_sample.median()
)

if not np.isfinite(
    FILTER_THRESHOLD
):
    FILTER_THRESHOLD = 0.0


print(
    "\nFilter condition:"
)

print(
    f"{FILTER_COL} > "
    f"{FILTER_THRESHOLD:,.6f}"
)


# Remove duplicate column names from read list.

USE_COLS = list(
    dict.fromkeys(
        [
            GROUP_COL,
            VALUE_COL,
            FILTER_COL
        ]
    )
)


# ================================================================
# 15. COMMON BENCHMARK TASK
#
# Every tool executes conceptually:
#
# SELECT
#    GROUP_COL,
#    COUNT(*),
#    SUM(VALUE_COL),
#    AVG(VALUE_COL)
#
# FROM parquet
#
# WHERE
#    FILTER_COL > threshold
#
# GROUP BY GROUP_COL
# ================================================================


# ================================================================
# 16. PANDAS
# ================================================================

def run_pandas():

    df = pd.read_parquet(
        PARQUET_PATH,
        columns=USE_COLS,
        engine="pyarrow"
    )

    df = df.dropna(
        subset=USE_COLS
    )

    df = df[
        df[FILTER_COL]
        > FILTER_THRESHOLD
    ]

    result = (

        df.groupby(
            GROUP_COL,
            sort=False
        )

        .agg(
            n_rows=(
                VALUE_COL,
                "size"
            ),

            sum_value=(
                VALUE_COL,
                "sum"
            ),

            mean_value=(
                VALUE_COL,
                "mean"
            )
        )

        .reset_index()
    )

    result = result.rename(
        columns={
            GROUP_COL:
            "group_key"
        }
    )

    return result


# ================================================================
# 17. POLARS
# ================================================================

def run_polars():

    q = (
        pl.scan_parquet(
            PARQUET_PATH
        )

        .select(
            USE_COLS
        )

        .filter(
            pl.col(GROUP_COL)
              .is_not_null()

            &

            pl.col(VALUE_COL)
              .is_not_null()

            &

            pl.col(FILTER_COL)
              .is_not_null()

            &

            (
                pl.col(FILTER_COL)
                > FILTER_THRESHOLD
            )
        )

        .group_by(
            GROUP_COL
        )

        .agg(

            pl.len()
              .alias(
                  "n_rows"
              ),

            pl.col(VALUE_COL)
              .sum()
              .alias(
                  "sum_value"
              ),

            pl.col(VALUE_COL)
              .mean()
              .alias(
                  "mean_value"
              )
        )
    )

    result = q.collect(
        engine="streaming"
    )

    result = (
        result
        .rename(
            {
                GROUP_COL:
                "group_key"
            }
        )
        .to_pandas()
    )

    return result


# ================================================================
# 18. DUCKDB
# ================================================================

def sql_identifier(name):

    return (
        '"'
        + name.replace(
            '"',
            '""'
        )
        + '"'
    )


def run_duckdb():

    g = sql_identifier(
        GROUP_COL
    )

    v = sql_identifier(
        VALUE_COL
    )

    f = sql_identifier(
        FILTER_COL
    )

    path_sql = (
        PARQUET_PATH
        .replace(
            "'",
            "''"
        )
    )

    con = duckdb.connect(
        database=":memory:"
    )

    con.execute(
        f"""
        SET threads={CPU_COUNT}
        """
    )

    query = f"""
    SELECT

        {g}
            AS group_key,

        COUNT(*)
            AS n_rows,

        SUM(
            CAST(
                {v}
                AS DOUBLE
            )
        )
            AS sum_value,

        AVG(
            CAST(
                {v}
                AS DOUBLE
            )
        )
            AS mean_value

    FROM
        read_parquet(
            '{path_sql}'
        )

    WHERE
        {g} IS NOT NULL

        AND
        {v} IS NOT NULL

        AND
        {f} IS NOT NULL

        AND
        CAST(
            {f}
            AS DOUBLE
        )
        >
        {FILTER_THRESHOLD}

    GROUP BY
        {g}
    """

    result = (
        con.execute(
            query
        )
        .fetchdf()
    )

    con.close()

    return result


# ================================================================
# 19. DASK
# ================================================================

def run_dask():

    df = dd.read_parquet(
        PARQUET_PATH,
        columns=USE_COLS
    )

    df = df.dropna(
        subset=USE_COLS
    )

    df = df[
        df[FILTER_COL]
        > FILTER_THRESHOLD
    ]

    result = (

        df.groupby(
            GROUP_COL
        )[VALUE_COL]

        .agg(
            [
                "count",
                "sum",
                "mean"
            ]
        )

        .compute(
            scheduler="threads",
            num_workers=CPU_COUNT
        )

        .reset_index()
    )

    result = result.rename(
        columns={
            GROUP_COL:
                "group_key",

            "count":
                "n_rows",

            "sum":
                "sum_value",

            "mean":
                "mean_value"
        }
    )

    return result


# ================================================================
# 20. VAEX
# ================================================================

VAEX_AVAILABLE = False

try:

    import vaex

    VAEX_AVAILABLE = True

    print(
        "\nVaex available."
    )

except Exception as e:

    print(
        "\nVaex unavailable:"
    )

    print(
        str(e)[:500]
    )


def run_vaex():

    df = vaex.open(
        PARQUET_PATH
    )

    # Remove missing/NaN values from
    # the benchmark columns.

    df = df.dropna(
        column_names=USE_COLS
    )

    df = df[
        df[FILTER_COL]
        > FILTER_THRESHOLD
    ]

    result = df.groupby(

        GROUP_COL,

        agg={

            "n_rows":
                vaex.agg.count(),

            "sum_value":
                vaex.agg.sum(
                    VALUE_COL
                ),

            "mean_value":
                vaex.agg.mean(
                    VALUE_COL
                )
        }
    )

    result = (
        result
        .to_pandas_df()
        .rename(
            columns={
                GROUP_COL:
                "group_key"
            }
        )
    )

    return result


# ================================================================
# 21. RAY DATA INITIALIZATION
#
# Initialization is deliberately OUTSIDE the timer.
# We measure data-processing time, not startup time.
# ================================================================

RAY_AVAILABLE = False
ray = None

try:

    import ray

    from ray.data.expressions import col
    from ray.data.aggregate import (
        Count,
        Sum,
        Mean
    )

    if ray.is_initialized():
        ray.shutdown()

    ray.init(
        num_cpus=CPU_COUNT,
        include_dashboard=False,
        log_to_driver=False,
        ignore_reinit_error=True
    )

    RAY_AVAILABLE = True

    print(
        "\nRay initialized."
    )

except Exception as e:

    print(
        "\nRay unavailable:"
    )

    print(
        str(e)[:500]
    )


def run_ray():

    # Ray workers use local:///
    # for an explicit local path.

    ray_path = (
        "local://"
        + PARQUET_PATH
    )

    ds = ray.data.read_parquet(
        ray_path,
        columns=USE_COLS
    )

    expression = (

        col(GROUP_COL)
        .is_not_null()

        &

        col(VALUE_COL)
        .is_not_null()

        &

        col(FILTER_COL)
        .is_not_null()

        &

        (
            col(FILTER_COL)
            > FILTER_THRESHOLD
        )
    )

    ds = ds.filter(
        expr=expression
    )

    result_ds = (

        ds.groupby(
            GROUP_COL
        )

        .aggregate(

            Count(
                alias_name=
                "n_rows"
            ),

            Sum(
                on=VALUE_COL,
                alias_name=
                "sum_value"
            ),

            Mean(
                on=VALUE_COL,
                alias_name=
                "mean_value"
            )
        )
    )

    result = pd.DataFrame(
        result_ds.take_all()
    )

    result = result.rename(
        columns={
            GROUP_COL:
            "group_key"
        }
    )

    return result


# ================================================================
# 22. PYSPARK INITIALIZATION
#
# Also outside benchmark timer.
# ================================================================

SPARK_AVAILABLE = False
spark = None

try:

    from pyspark.sql import (
        SparkSession
    )

    from pyspark.sql import (
        functions as F
    )

    spark = (

        SparkSession
        .builder

        .master(
            f"local[{CPU_COUNT}]"
        )

        .appName(
            "BigDataBenchmark"
        )

        .config(
            "spark.ui.enabled",
            "false"
        )

        .config(
            "spark.sql.shuffle.partitions",
            str(
                max(
                    4,
                    CPU_COUNT * 2
                )
            )
        )

        .getOrCreate()
    )

    spark.sparkContext.setLogLevel(
        "ERROR"
    )

    SPARK_AVAILABLE = True

    print(
        "\nPySpark initialized."
    )

except Exception as e:

    print(
        "\nPySpark unavailable:"
    )

    print(
        str(e)[:500]
    )


def run_pyspark():

    df = (

        spark.read
        .parquet(
            PARQUET_PATH
        )

        .select(
            *USE_COLS
        )
    )

    df = df.filter(

        F.col(
            GROUP_COL
        ).isNotNull()

        &

        F.col(
            VALUE_COL
        ).isNotNull()

        &

        F.col(
            FILTER_COL
        ).isNotNull()

        &

        (
            F.col(
                FILTER_COL
            )
            >
            F.lit(
                FILTER_THRESHOLD
            )
        )
    )

    result = (

        df.groupBy(
            GROUP_COL
        )

        .agg(

            F.count(
                F.lit(1)
            ).alias(
                "n_rows"
            ),

            F.sum(
                F.col(
                    VALUE_COL
                ).cast(
                    "double"
                )
            ).alias(
                "sum_value"
            ),

            F.avg(
                F.col(
                    VALUE_COL
                ).cast(
                    "double"
                )
            ).alias(
                "mean_value"
            )
        )

        .withColumnRenamed(
            GROUP_COL,
            "group_key"
        )

        .toPandas()
    )

    return result


# ================================================================
# 23. REGISTER AVAILABLE ENGINES
# ================================================================

TOOLS = {
    "Pandas":
        run_pandas,

    "Polars":
        run_polars,

    "DuckDB":
        run_duckdb,

    "Dask":
        run_dask,
}


if VAEX_AVAILABLE:

    TOOLS[
        "Vaex"
    ] = run_vaex


if RAY_AVAILABLE:

    TOOLS[
        "Ray Data"
    ] = run_ray


if SPARK_AVAILABLE:

    TOOLS[
        "PySpark"
    ] = run_pyspark


print("\n" + "=" * 80)
print("TOOLS INCLUDED IN BENCHMARK")
print("=" * 80)

for tool in TOOLS:
    print(
        " •",
        tool
    )


# ================================================================
# 24. BENCHMARK
#
# The order is randomized each round to reduce
# systematic cache-order bias.
#
# What is timed:
#
#   Read Parquet
#       +
#   Projection
#       +
#   Filter
#       +
#   GroupBy
#       +
#   Count/Sum/Mean
#       +
#   Materialize result
#
# What is NOT timed:
#
#   pip installation
#   Ray startup
#   Spark startup
# ================================================================

timings = {
    tool: []
    for tool in TOOLS
}

last_outputs = {}

errors = {}


print("\n" + "=" * 80)
print("STARTING BENCHMARK")
print("=" * 80)


for run_idx in range(
    N_RUNS
):

    print(
        f"\n===== ROUND "
        f"{run_idx + 1}"
        f"/{N_RUNS} ====="
    )

    tool_order = list(
        TOOLS.items()
    )

    rng = random.Random(
        RANDOM_SEED
        + run_idx
    )

    rng.shuffle(
        tool_order
    )

    for tool_name, function in tool_order:

        if tool_name in errors:
            continue

        gc.collect()

        try:

            start = (
                time.perf_counter()
            )

            output = function()

            elapsed = (
                time.perf_counter()
                - start
            )

            timings[
                tool_name
            ].append(
                elapsed
            )

            last_outputs[
                tool_name
            ] = output

            print(
                f"{tool_name:12s}"
                f" : "
                f"{elapsed:,.4f} sec"
            )

        except Exception as e:

            errors[
                tool_name
            ] = str(e)

            print(
                f"{tool_name:12s}"
                f" : ERROR"
            )

            print(
                str(e)[:1000]
            )


# ================================================================
# 25. CREATE SUMMARY TABLE
# ================================================================

summary_rows = []

for tool_name, values in timings.items():

    if not values:
        continue

    summary_rows.append(
        {
            "Tool":
                tool_name,

            "Runs":
                len(values),

            "First_Run_s":
                values[0],

            "Best_s":
                np.min(
                    values
                ),

            "Median_s":
                np.median(
                    values
                ),

            "Mean_s":
                np.mean(
                    values
                ),

            "Std_s":
                np.std(
                    values
                )
        }
    )


results_df = pd.DataFrame(
    summary_rows
)

if results_df.empty:

    raise RuntimeError(
        "No benchmark tool completed successfully."
    )


# Ranking uses MEDIAN time.

results_df = (
    results_df
    .sort_values(
        "Median_s",
        ascending=True
    )
    .reset_index(
        drop=True
    )
)


results_df.insert(
    0,
    "Rank",
    np.arange(
        1,
        len(results_df) + 1
    )
)


# ================================================================
# 26. SPEEDUP VS PANDAS
# ================================================================

if (
    results_df["Tool"]
    == "Pandas"
).any():

    pandas_time = float(
        results_df.loc[
            results_df["Tool"]
            == "Pandas",
            "Median_s"
        ].iloc[0]
    )

    results_df[
        "Speedup_vs_Pandas_x"
    ] = (
        pandas_time
        /
        results_df[
            "Median_s"
        ]
    )

else:

    results_df[
        "Speedup_vs_Pandas_x"
    ] = np.nan


# Round display values.

display_df = (
    results_df.copy()
)

for c in [
    "First_Run_s",
    "Best_s",
    "Median_s",
    "Mean_s",
    "Std_s",
    "Speedup_vs_Pandas_x"
]:

    display_df[c] = (
        display_df[c]
        .round(4)
    )


print("\n" + "=" * 80)
print("FINAL PERFORMANCE RANKING")
print("=" * 80)

display(
    display_df
)


# ================================================================
# 27. FASTEST TOOL
# ================================================================

FASTEST_TOOL = (
    results_df
    .iloc[0][
        "Tool"
    ]
)

FASTEST_TIME = float(
    results_df
    .iloc[0][
        "Median_s"
    ]
)


print("\n" + "=" * 80)
print("WINNER")
print("=" * 80)

print(
    "🏆 FASTEST TOOL:",
    FASTEST_TOOL
)

print(
    f"Median execution time:"
    f" {FASTEST_TIME:,.4f} sec"
)


if (
    results_df["Tool"]
    == "Pandas"
).any():

    speedup = float(
        results_df
        .iloc[0][
            "Speedup_vs_Pandas_x"
        ]
    )

    print(
        f"Speedup vs Pandas:"
        f" {speedup:,.2f}x"
    )


# ================================================================
# 28. VALIDATE RESULTS
#
# We check:
#   - number of groups
#   - number of filtered rows
#   - global sum
#
# Small floating point differences are allowed.
# ================================================================

def output_signature(
    df
):

    x = df.copy()

    n_groups = len(
        x
    )

    n_rows = int(
        pd.to_numeric(
            x["n_rows"],
            errors="coerce"
        )
        .fillna(0)
        .sum()
    )

    total_sum = float(
        pd.to_numeric(
            x["sum_value"],
            errors="coerce"
        )
        .fillna(0)
        .sum()
    )

    return {
        "Groups":
            n_groups,

        "Filtered_rows":
            n_rows,

        "Total_sum":
            total_sum
    }


validation_rows = []

for tool_name, df_result in last_outputs.items():

    try:

        sig = output_signature(
            df_result
        )

        validation_rows.append(
            {
                "Tool":
                    tool_name,

                **sig
            }
        )

    except Exception:
        pass


validation_df = pd.DataFrame(
    validation_rows
)


print("\n" + "=" * 80)
print("RESULT VALIDATION")
print("=" * 80)

display(
    validation_df
)


# ================================================================
# 29. VALIDATION AGAINST PANDAS
# ================================================================

if (
    "Pandas"
    in last_outputs
):

    reference = output_signature(
        last_outputs[
            "Pandas"
        ]
    )

    checks = []

    for tool_name, df_result in last_outputs.items():

        sig = output_signature(
            df_result
        )

        same_groups = (
            sig["Groups"]
            ==
            reference["Groups"]
        )

        same_rows = (
            sig["Filtered_rows"]
            ==
            reference["Filtered_rows"]
        )

        same_sum = np.isclose(

            sig["Total_sum"],

            reference[
                "Total_sum"
            ],

            rtol=1e-5,
            atol=1e-5
        )

        checks.append(
            {
                "Tool":
                    tool_name,

                "Same_groups":
                    same_groups,

                "Same_rows":
                    same_rows,

                "Same_sum":
                    bool(
                        same_sum
                    ),

                "VALID":
                    (
                        same_groups
                        and same_rows
                        and same_sum
                    )
            }
        )

    check_df = pd.DataFrame(
        checks
    )

    print(
        "\nConsistency with Pandas:"
    )

    display(
        check_df
    )


# ================================================================
# 30. BAR CHART:
#     LOWER TIME = BETTER
# ================================================================

plot_df = (
    results_df
    .sort_values(
        "Median_s",
        ascending=True
    )
)


plt.figure(
    figsize=(10, 6)
)

bars = plt.barh(
    plot_df["Tool"],
    plot_df["Median_s"]
)

plt.gca().invert_yaxis()

plt.xlabel(
    "Median execution time (seconds)"
)

plt.ylabel(
    "Big Data tool"
)

plt.title(
    "Parquet Big Data Processing Performance"
)

for bar, value in zip(
    bars,
    plot_df["Median_s"]
):

    plt.text(
        value,
        bar.get_y()
        + bar.get_height() / 2,

        f" {value:.3f}s",

        va="center"
    )

plt.tight_layout()

plt.show()


# ================================================================
# 31. SPEEDUP TABLE
# ================================================================

speed_df = results_df[
    [
        "Rank",
        "Tool",
        "Median_s",
        "Speedup_vs_Pandas_x"
    ]
].copy()


speed_df[
    "Median_s"
] = (
    speed_df[
        "Median_s"
    ].round(4)
)


speed_df[
    "Speedup_vs_Pandas_x"
] = (
    speed_df[
        "Speedup_vs_Pandas_x"
    ].round(2)
)


print("\n" + "=" * 80)
print("SPEEDUP VS PANDAS")
print("=" * 80)

display(
    speed_df
)


# ================================================================
# 32. SHOW TOP 10 ANALYTIC RESULTS
#     PRODUCED BY THE FASTEST TOOL
# ================================================================

fast_result = (
    last_outputs[
        FASTEST_TOOL
    ]
    .copy()
)


fast_result[
    "sum_value"
] = pd.to_numeric(
    fast_result[
        "sum_value"
    ],
    errors="coerce"
)


fast_result = (
    fast_result
    .sort_values(
        "sum_value",
        ascending=False
    )
)


print(
    "\nTop 10 groups generated by:",
    FASTEST_TOOL
)

display(
    fast_result.head(10)
)


# ================================================================
# 33. SAVE BENCHMARK RESULT
# ================================================================

OUTPUT_FILE = (
    "/content/"
    "bigdata_parquet_benchmark_results.csv"
)

results_df.to_csv(
    OUTPUT_FILE,
    index=False
)


print("\n" + "=" * 80)
print("FINISHED")
print("=" * 80)

print(
    "Benchmark result saved to:"
)

print(
    OUTPUT_FILE
)


# ================================================================
# 34. INTERPRETATION
# ================================================================

print(
    f"""
============================================================
BENCHMARK INTERPRETATION
============================================================

Dataset
-------
Rows       : {NUM_ROWS:,}
Columns    : {NUM_COLS:,}
Size       : {FILE_SIZE_MB:,.2f} MB
Row groups : {NUM_ROW_GROUPS:,}

Analysis
--------
GROUP BY    : {GROUP_COL}
VALUE       : {VALUE_COL}
FILTER      : {FILTER_COL}

Condition
---------
{FILTER_COL} > {FILTER_THRESHOLD:,.6f}

Winner
------
{FASTEST_TOOL}

Median time
-----------
{FASTEST_TIME:,.4f} seconds

IMPORTANT
---------
This result means:

"{FASTEST_TOOL} was fastest for THIS Parquet file,
THIS analytical workload, and THIS Colab machine."

It does NOT automatically mean that the tool is fastest
for every Big Data workload.

Ray / Spark / Dask are particularly important when the
workload scales from one machine to distributed clusters.

Polars / DuckDB are often particularly competitive for
single-machine analytical processing.

Vaex focuses strongly on out-of-core and memory-efficient
processing.

============================================================
"""
)


# ================================================================
# OPTIONAL CLEANUP
# ================================================================

# Uncomment when completely finished:
#
# if RAY_AVAILABLE:
#     ray.shutdown()
#
# if SPARK_AVAILABLE:
#     spark.stop()
```
